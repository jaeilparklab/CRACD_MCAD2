```python
import numpy as np
import pandas as pd
import scanpy as sc
import scanpy.external as sce
import scipy
sc.settings.verbosity = 3
sc.logging.print_header()
sc.set_figure_params(dpi=100, dpi_save=600)

import os
```

```python
#### GSE236698

# Define sample list for batch processing
samples = ['P1_N', 'P2_N', 'P3_N', 'P4_N', 'P5_N', 'P6_N',
           'P1_T', 'P2_T', 'P3_T', 'P4_T', 'P5_T', 'P6_T']

# Base directory for all samples
base_path = "/mnt/e/GSE236698/Matrix/scRNA-seq/"

# Process each sample sequentially
for sample in samples:
    print(f"Processing sample: {sample}")
    
    # Read 10X matrix data preprocessed by SoupX
    adata = sc.read_10x_mtx(
        path=f"{base_path}{sample}/outs/filtered_feature_bc_matrix_SoupX/",
        var_names='gene_symbols', 
        cache=True  # Cache data for faster subsequent access
    )
    
    # Ensure gene names are unique
    adata.var_names_make_unique()
    
    sce.pp.scrublet(adata, sim_doublet_ratio=1.5)
    adata = adata[adata.obs['predicted_doublet'] == False]
   
    # Save processed data with compression
    output_path = f"{base_path}{sample}/outs/{sample}_SoupX_Scrublet.h5ad"
    adata.write(output_path, compression='gzip')
    print(f"Saved processed data to: {output_path}\n")

print("Batch processing completed for all samples!")
```
